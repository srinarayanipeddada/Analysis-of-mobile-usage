# analysis-checkpoint.ipynb (fixed)

This notebook was cleaned to: call `df.info()` correctly, coerce bbox columns to numeric, fill numeric NaNs, prevent negative box dimensions, and limit correlation to numeric columns.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# --- Load CSV (adjust path if needed) ---
path = "labels.csv"
if not os.path.exists(path):
    raise FileNotFoundError(f"'{path}' not found. Please put labels.csv in the working directory or change the path.")

df = pd.read_csv(path)

# --- Quick overview ---
print("Shape:", df.shape)
print("
First rows:")
print(df.head())

print("
Info:")
df.info()
print("
Describe:")
print(df.describe(include="all"))

# --- Ensure bounding box columns are numeric ---
bbox_cols = ["xmin", "xmax", "ymin", "ymax"]
missing_cols = [c for c in bbox_cols if c not in df.columns]
if missing_cols:
    raise KeyError(f"Missing expected bbox columns: {missing_cols}")

# convert to numeric (coerce errors to NaN)
df[bbox_cols] = df[bbox_cols].apply(pd.to_numeric, errors="coerce")

# report nulls before filling
print("
Null counts before filling:")
print(df.isnull().sum())

# Fill numeric NaNs with column means (affects only numeric columns)
numeric_means = df.select_dtypes(include=[np.number]).mean()
df.fillna(numeric_means, inplace=True)

# Confirm nulls afterwards (non-numeric NaNs might remain)
print("
Null counts after filling numeric columns:")
print(df.isnull().sum())

# --- Compute widths/heights/area robustly ---
# Prevent negative widths/heights if xmax < xmin due to bad data by clipping to 0
df["box_width"] = (df["xmax"] - df["xmin"]).clip(lower=0)
df["box_height"] = (df["ymax"] - df["ymin"]).clip(lower=0)
df["box_area"] = df["box_width"] * df["box_height"]

# --- Class distribution plot (if 'class' exists) ---
if "class" in df.columns:
    plt.figure(figsize=(6, 4))
    sns.countplot(x="class", data=df)
    plt.title("Object Count by Class")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("No 'class' column found in dataframe. Skipping class count plot.")

# --- Bounding box area distribution ---
plt.figure(figsize=(8, 5))
sns.histplot(df["box_area"].dropna(), bins=30, kde=False)
plt.title("Bounding Box Area Distribution")
plt.xlabel("Area")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# Print average dimensions
print("Mean box width: {:.3f}".format(df["box_width"].mean()))
print("Mean box height: {:.3f}".format(df["box_height"].mean()))

# --- Correlation heatmap for numeric columns only ---
numeric_df = df.select_dtypes(include=[np.number])
if numeric_df.shape[1] >= 2:
    corr = numeric_df.corr()
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Correlation Matrix (numeric columns)")
    plt.tight_layout()
    plt.show()
else:
    print("Not enough numeric columns to compute a correlation matrix.")

# --- Width vs Height scatter ---
plt.figure(figsize=(6, 5))
sns.scatterplot(x="box_width", y="box_height", data=df, alpha=0.6)
plt.title("Bounding Box Width vs Height")
plt.xlabel("Width")
plt.ylabel("Height")
plt.tight_layout()
plt.show()
```
